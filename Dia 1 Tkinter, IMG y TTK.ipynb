```python
# 02_label_propiedades.py
import tkinter as tk
from tkinter import font


root = tk.Tk()
root.title("Modificar propiedades de Label")
root.geometry("420x220")


# Fuente personalizada
fuente = font.Font(family="Segoe UI", size=14, weight="bold")


label = tk.Label(root, text="Pasa el ratón y cambia el estilo", width=35, anchor="center")
label.grid(row=0, column=0, columnspan=3, pady=12)


# Botones para cambiar propiedades


def set_color_fg():
    label.config(fg="blue")


def set_color_bg():
    label.config(bg="#f0f8ff")


def set_fuente():
    label.config(font=fuente)


def reset():
    label.config(text="Pasa el ratón y cambia el estilo", fg="black", bg=root.cget("bg"), font=("Segoe UI", 11))


btn_fg = tk.Button(root, text="Texto azul", command=set_color_fg)
btn_bg = tk.Button(root, text="Fondo claro", command=set_color_bg)
btn_font = tk.Button(root, text="Fuente bold", command=set_fuente)
btn_reset = tk.Button(root, text="Reset", command=reset)


btn_fg.grid(row=1, column=0, padx=5, pady=8)
btn_bg.grid(row=1, column=1, padx=5, pady=8)
btn_font.grid(row=1, column=2, padx=5, pady=8)
btn_reset.grid(row=2, column=0, columnspan=3, pady=8)


# Eventos de ratón


def on_enter(_):
    label.config(text="¡Dentro del Label!", underline=0)


def on_leave(_):
    label.config(text="Pasa el ratón y cambia el estilo", underline=-1)


label.bind("<Enter>", on_enter)
label.bind("<Leave>", on_leave)


root.mainloop()
```

4) Mini calculadora

```python
# 04_mini_calculadora.py
import tkinter as tk
from tkinter import messagebox


root = tk.Tk()
root.title("Mini Calculadora")
root.geometry("290x360")
root.resizable(False, False)


expresion = tk.StringVar(value="")


entry = tk.Entry(root, textvariable=expresion, justify="right", font=("Consolas", 16))
entry.grid(row=0, column=0, columnspan=4, padx=10, pady=10, sticky="nsew")


# Configurar rejilla para que crezca
for i in range(4):
    root.grid_columnconfigure(i, weight=1)
for i in range(6):
    root.grid_rowconfigure(i, weight=1)


# Botones
botones = [
("7", 1, 0), ("8", 1, 1), ("9", 1, 2), ("/", 1, 3),
("4", 2, 0), ("5", 2, 1), ("6", 2, 2), ("*", 2, 3),
("1", 3, 0), ("2", 3, 1), ("3", 3, 2), ("-", 3, 3),
("0", 4, 0), (".", 4, 1), ("C", 4, 2), ("+", 4, 3),
]


def agregar(char):
    actual = expresion.get()
    expresion.set(actual + char)


for (txt, r, c) in botones:
    if txt == "C":
        def make_clear():
            return lambda: expresion.set("")
        cmd = make_clear()
    else:
        cmd = lambda ch=txt: agregar(ch)
    tk.Button(root, text=txt, command=cmd, width=4, height=2).grid(row=r, column=c, padx=5, pady=5, sticky="nsew")


# Botón igual ocupa toda la última fila


def calcular():
    try:
        expr = expresion.get()
        # Seguridad básica: solo permitir dígitos, operadores y punto
        for ch in expr:
            if ch not in "0123456789.+-*/() ":
                raise ValueError("Carácter no permitido")
            resultado = eval(expr)
            expresion.set(str(resultado))
    except Exception as e:
        messagebox.showerror("Error", f"Expresión inválida:\n{e}")


btn_igual = tk.Button(root, text="=", command=calcular, height=2)
btn_igual.grid(row=5, column=0, columnspan=4, padx=5, pady=5, sticky="nsew")


root.mainloop()
```

4a Calculadora sin lambda

```python
# 04_mini_calculadora.py
import tkinter as tk
from tkinter import messagebox


root = tk.Tk()
root.title("Mini Calculadora")
root.geometry("290x360")
root.resizable(False, False)


expresion = tk.StringVar(value="")


entry = tk.Entry(root, textvariable=expresion, justify="right", font=("Consolas", 16))
entry.grid(row=0, column=0, columnspan=4, padx=10, pady=10, sticky="nsew")


# Configurar rejilla para que crezca
for i in range(4):
    root.grid_columnconfigure(i, weight=1)
for i in range(6):
    root.grid_rowconfigure(i, weight=1)


# Definimos botones (texto, fila, columna)
botones = [
    ("7", 1, 0), ("8", 1, 1), ("9", 1, 2), ("/", 1, 3),
    ("4", 2, 0), ("5", 2, 1), ("6", 2, 2), ("*", 2, 3),
    ("1", 3, 0), ("2", 3, 1), ("3", 3, 2), ("-", 3, 3),
    ("0", 4, 0), (".", 4, 1), ("C", 4, 2), ("+", 4, 3),
]

def agregar(char):
    actual = expresion.get()
    expresion.set(actual + char)

# Función que devuelve un "command" para cada botón normal
def crear_comando(char):
    def comando():
        agregar(char)
    return comando

# Función especial para limpiar
def comando_clear():
    expresion.set("")

# Crear botones
for (txt, r, c) in botones:
    if txt == "C":
        cmd = comando_clear
    else:
        cmd = crear_comando(txt)
    tk.Button(root, text=txt, command=cmd, width=4, height=2).grid(
        row=r, column=c, padx=5, pady=5, sticky="nsew"
    )


# Botón igual ocupa toda la última fila


def calcular():
    try:
        expr = expresion.get()
        # Seguridad básica: solo permitir dígitos, operadores y punto
        for ch in expr:
            if ch not in "0123456789.+-*/() ":
                raise ValueError("Carácter no permitido")
            resultado = eval(expr)
            expresion.set(str(resultado))
    except Exception as e:
        messagebox.showerror("Error", f"Expresión inválida:\n{e}")


btn_igual = tk.Button(root, text="=", command=calcular, height=2)
btn_igual.grid(row=5, column=0, columnspan=4, padx=5, pady=5, sticky="w")


root.mainloop()
```

Aqui generamos una tabla sencilla
Treeview es un widget genérico que sirve tanto como:

Árbol (jerarquías, como un explorador de archivos).

Tabla (cuando usas show="headings").

Por defecto es un árbol: cada fila puede tener hijos, que se expanden o colapsan.
Cuando lo usas como tabla, realmente estás aprovechando ese árbol, pero restringido a nivel raíz (sin hijos)

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Treeview")
# Usamos treeview para generar tablas
# Permite ordenar, seleccionar filas, etc. Al usar show="headings" se comporta como tabla
# Si no usas show="headings" se comporta como árbol (con jerarquías)
tree = ttk.Treeview(root, columns=("Nombre", "Edad"), show="headings")
tree.heading("Nombre", text="Nombre")
tree.heading("Edad", text="Edad")

# Al hacer la inserción, el primer parámetro es el padre ("" es raíz)
# El segundo es la posición (end para añadir al final)
# El tercero son los valores de las columnas
tree.insert("", "end", values=("Ana", 25))
tree.insert("", "end", values=("Luis", 30))

tree.pack()
root.mainloop()
```

En este ejemplo ademas de la tabla haremos selección de elementos y ordenación

```python
import tkinter as tk
from tkinter import ttk

def sort_column(tree, col, reverse):
    # Obtener datos actuales de la columna, se guardan en un array de tuplas (valor, item), 
    # donde item es el ID del elemento, guardando algo como ('Ana', 'I001'), y para la edad ('25', 'I001'), 
    # tree.set obtiene el valor de la columna col para el item, pero puede dar valor tambien, por ejemplo 
    # Obtener un valor de una celda: nombre = tree.set(item_id, "Nombre")
    # Dar valor a una celda: tree.set(item_id, "Edad", 26)
    data = [(tree.set(item, col), item) for item in tree.get_children('')]

    # Intentar convertir a número si es posible (para ordenar números correctamente)
    try:
        data = [(float(val), item) for val, item in data]
    except ValueError:
        pass

    # Ordenar datos, reverse=reverse para alternar entre ascendente/descendente
    data.sort(reverse=reverse)

    # Reordenar elementos en el Treeview, tree.move mueve el item a la posición index, y '' es la raíz
    for index, (val, item) in enumerate(data):
        tree.move(item, '', index)

    # Alternar el orden para el próximo clic
    tree.heading(col, command=lambda: sort_column(tree, col, not reverse))

# --- Ventana principal ---
root = tk.Tk()
root.title("Treeview Ordenable")

# Crear Treeview
tree = ttk.Treeview(root, columns=("Nombre", "Edad"), show="headings")
tree.pack(padx=20, pady=20)

# Configurar encabezados, hay que usar lambda para pasar parámetros, 
# si no pasamos parametros podemos usar una función normal
tree.heading("Nombre", text="Nombre", command=lambda: sort_column(tree, "Nombre", False))
tree.heading("Edad", text="Edad", command=lambda: sort_column(tree, "Edad", False))

# Insertar datos
tree.insert("", "end", values=("Ana", 25))
tree.insert("", "end", values=("Luis", 30))
tree.insert("", "end", values=("Pedro", 20))
tree.insert("", "end", values=("Sofía", 28))

root.mainloop()
```

Vamos a añadir un formulario

```python
import tkinter as tk
from tkinter import ttk, messagebox

# ---------- Funciones de orden ----------
def sort_column(tree, col, reverse):
    data = [(tree.set(item, col), item) for item in tree.get_children('')]
    # Intentar ordenar numéricamente si procede
    try:
        data = [(float(v), i) for v, i in data]
    except ValueError:
        pass
    data.sort(reverse=reverse)
    for idx, (_, item) in enumerate(data):
        tree.move(item, '', idx)
    tree.heading(col, command=lambda: sort_column(tree, col, not reverse))

# ---------- Funciones del CRUD ----------
def clear_form():
    nombre_var.set("")
    edad_var.set("")
    tree.selection_remove(tree.selection())

def validate_inputs():
    nombre = nombre_var.get().strip()
    edad_txt = edad_var.get().strip()
    if not nombre:
        messagebox.showwarning("Validación", "El nombre no puede estar vacío.")
        return None
    if not edad_txt.isdigit():
        messagebox.showwarning("Validación", "La edad debe ser un número entero no negativo.")
        return None
    return nombre, int(edad_txt)

def add_record():
    vals = validate_inputs()
    if not vals:
        return
    nombre, edad = vals
    tree.insert("", "end", values=(nombre, edad))
    clear_form()

def update_record():
    sel = tree.selection()
    if not sel:
        messagebox.showinfo("Modificar", "Selecciona una fila para modificar.")
        return
    vals = validate_inputs()
    if not vals:
        return
    nombre, edad = vals
    item_id = sel[0]
    tree.item(item_id, values=(nombre, edad))
    clear_form()

def delete_record():
    sel = tree.selection()
    if not sel:
        messagebox.showinfo("Eliminar", "Selecciona una fila para eliminar.")
        return
    if messagebox.askyesno("Confirmar eliminación", "¿Seguro que quieres eliminar la fila seleccionada?"):
        tree.delete(sel[0])
        clear_form()

def on_select(event):
    sel = tree.selection()
    if not sel:
        return
    vals = tree.item(sel[0], "values")
    if vals:
        nombre_var.set(vals[0])
        edad_var.set(vals[1])

# ---------- Ventana principal ----------
root = tk.Tk()
root.title("Treeview con formulario (CRUD)")

# Frame contenedor con padding respecto a la ventana
container = ttk.Frame(root, padding=20)
container.pack(fill="both", expand=True)

# ---------- Treeview ----------
tree = ttk.Treeview(container, columns=("Nombre", "Edad"), show="headings", height=8)
tree.grid(row=0, column=0, columnspan=4, sticky="nsew")

# Encabezados clicables (orden)
tree.heading("Nombre", text="Nombre", command=lambda: sort_column(tree, "Nombre", False))
tree.heading("Edad", text="Edad", command=lambda: sort_column(tree, "Edad", False))

# Ancho de columnas
tree.column("Nombre", width=180, anchor="w")
tree.column("Edad", width=80, anchor="center")

# Scrollbar vertical
scrollbar = ttk.Scrollbar(container, orient="vertical", command=tree.yview)
scrollbar.grid(row=0, column=4, sticky="ns")
tree.configure(yscrollcommand=scrollbar.set)

# Datos iniciales
for fila in [("Ana", 25), ("Luis", 30), ("Pedro", 20), ("Sofía", 28)]:
    tree.insert("", "end", values=fila)

# Evento de selección para autocompletar el formulario
tree.bind("<<TreeviewSelect>>", on_select)

# Permitir que el treeview se expanda
container.rowconfigure(0, weight=1)
container.columnconfigure(0, weight=1)

# ---------- Formulario ----------
nombre_var = tk.StringVar()
edad_var = tk.StringVar()

ttk.Label(container, text="Nombre:").grid(row=1, column=0, sticky="w", pady=(12, 4))
nombre_entry = ttk.Entry(container, textvariable=nombre_var, width=25)
nombre_entry.grid(row=1, column=1, sticky="ew", pady=(12, 4))

ttk.Label(container, text="Edad:").grid(row=1, column=2, sticky="w", pady=(12, 4))
edad_entry = ttk.Entry(container, textvariable=edad_var, width=10)
edad_entry.grid(row=1, column=3, sticky="w", pady=(12, 4))

# ---------- Botones ----------
btn_frame = ttk.Frame(container)
btn_frame.grid(row=2, column=0, columnspan=4, sticky="ew", pady=(8, 0))
btn_frame.columnconfigure((0,1,2,3), weight=1)

ttk.Button(btn_frame, text="Añadir", command=add_record).grid(row=0, column=0, padx=5, pady=8, sticky="ew")
ttk.Button(btn_frame, text="Modificar", command=update_record).grid(row=0, column=1, padx=5, pady=8, sticky="ew")
ttk.Button(btn_frame, text="Eliminar", command=delete_record).grid(row=0, column=2, padx=5, pady=8, sticky="ew")
ttk.Button(btn_frame, text="Limpiar", command=clear_form).grid(row=0, column=3, padx=5, pady=8, sticky="ew")

# Foco inicial
nombre_entry.focus()

root.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

def on_destroy(event):
    widget = event.widget
    print(f"Se ha destruido el widget: {widget}")
    
    # Si el widget destruido es la ventana principal
    if widget == root:
        messagebox.showinfo("Adiós", "¡La ventana se ha cerrado!")

# ---------------------------
# Ventana principal
# ---------------------------
root = tk.Tk()
root.title("Ejemplo bind <Destroy>")
root.geometry("300x200")

# ---------------------------
# Widgets
# ---------------------------
label = tk.Label(root, text="Cierra la ventana para activar el evento.")
label.pack(pady=20)

btn = tk.Button(root, text="Cerrar ventana", command=root.destroy)
btn.pack(pady=10)

# ---------------------------
# Bind al evento <Destroy>
# ---------------------------
root.bind("<Destroy>", on_destroy)

root.mainloop()
```
